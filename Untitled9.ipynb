```python
import pandas as pd
import os

file_path = os.path.join(path, "dataset-tickets-multi-lang-4-20k.csv")

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Create a copy of the original dataset
data = df.copy()

# Remove rows where the ticket body is missing
data = data.dropna(subset=["body"])

# Fill missing subjects with an empty string
data["subject"] = data["subject"].fillna("")

# Combine subject and body into one text field
data["text"] = data["subject"] + " " + data["body"]

# Remove unnecessary columns
data = data.drop(
    columns=["answer", "tag_1", "tag_2", "tag_3", "tag_4",
             "tag_5", "tag_6", "tag_7", "tag_8"]
)

print("Cleaned dataset shape:", data.shape)
print("\nColumns:")
print(data.columns.tolist())
```

```python
# Create English-only dataset for the first NLP model
english_data = data[data["language"] == "en"].copy()

print("English dataset shape:", english_data.shape)

print("\nCategory distribution:")
print(english_data["queue"].value_counts())

print("\nPriority distribution:")
print(english_data["priority"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

# Input text
X = english_data["text"]

# Target variable
y = english_data["queue"]

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=10000,
    ngram_range=(1, 2)
)

# Learn vocabulary from training data and transform it
X_train_tfidf = tfidf.fit_transform(X_train)

# Transform test data using the same vocabulary
X_test_tfidf = tfidf.transform(X_test)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
tfidf.fit_transform(X_train)
tfidf.transform(X_test)
```

```python
print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create the baseline model
model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
model.fit(X_train_tfidf, y_train)

print("Model training completed!")
```

```python
# Predict the queue for the test tickets
y_pred = model.predict(X_test_tfidf)

print("Predictions completed!")
print("\nFirst 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.svm import LinearSVC

# Create Linear SVM model
svm_model = LinearSVC(
    random_state=42
)

# Train the model
svm_model.fit(X_train_tfidf, y_train)

print("Linear SVM training completed!")
```

```python
# Make predictions using Linear SVM
svm_pred = svm_model.predict(X_test_tfidf)

print("SVM predictions completed!")
print("\nFirst 10 predictions:")
print(svm_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, classification_report, f1_score

# Calculate accuracy
svm_accuracy = accuracy_score(y_test, svm_pred)

# Calculate Macro F1
svm_macro_f1 = f1_score(y_test, svm_pred, average="macro")

print("SVM Accuracy:", svm_accuracy)
print("SVM Macro F1:", svm_macro_f1)

print("\nSVM Classification Report:")
print(classification_report(
    y_test,
    svm_pred,
    zero_division=0
))
```

```python
from sklearn.model_selection import train_test_split

# Split training data into training and validation sets
X_train_final, X_val, y_train_final, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Final training samples:", len(X_train_final))
print("Validation samples:", len(X_val))
print("Test samples:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF vectorizer
tfidf_final = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=20000,
    ngram_range=(1, 2),
    sublinear_tf=True
)

# Fit ONLY on training data
X_train_final_tfidf = tfidf_final.fit_transform(X_train_final)

# Transform validation and test data
X_val_tfidf = tfidf_final.transform(X_val)
X_test_final_tfidf = tfidf_final.transform(X_test)

print("Final training TF-IDF shape:", X_train_final_tfidf.shape)
print("Validation TF-IDF shape:", X_val_tfidf.shape)
print("Final test TF-IDF shape:", X_test_final_tfidf.shape)
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score

C_values = [0.5, 1.0, 2.0, 3.0]

validation_results = []

for c in C_values:
    svm_val = LinearSVC(
        C=c,
        class_weight="balanced",
        random_state=42
    )

    # Train only on final training data
    svm_val.fit(
        X_train_final_tfidf,
        y_train_final
    )

    # Predict validation data
    val_pred = svm_val.predict(X_val_tfidf)

    # Evaluate validation performance
    val_accuracy = accuracy_score(y_val, val_pred)
    val_macro_f1 = f1_score(
        y_val,
        val_pred,
        average="macro"
    )

    validation_results.append(
        [c, val_accuracy, val_macro_f1]
    )

print("Validation tuning results:\n")

for result in validation_results:
    print(
        f"C = {result[0]} | "
        f"Accuracy = {result[1]:.4f} | "
        f"Macro F1 = {result[2]:.4f}"
    )
```

```python
# Final SVM model selected using validation Macro F1
final_model = LinearSVC(
    C=2.0,
    class_weight="balanced",
    random_state=42
)

# Train on all training data
final_model.fit(
    X_train_tfidf,
    y_train
)

print("Final queue classification model trained!")
```

```python
# Final predictions on the untouched test set
final_pred = final_model.predict(X_test_tfidf)

print("Final predictions completed!")
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report

final_accuracy = accuracy_score(y_test, final_pred)

final_macro_f1 = f1_score(
    y_test,
    final_pred,
    average="macro"
)

print("Final Test Accuracy:", final_accuracy)
print("Final Test Macro F1:", final_macro_f1)

print("\nFinal Classification Report:")
print(classification_report(
    y_test,
    final_pred,
    zero_division=0
))
```

```python
from sklearn.model_selection import train_test_split

# Text input and priority target
X_priority = english_data["text"]
y_priority = english_data["priority"]

# First split: Training + Test
X_priority_train, X_priority_test, y_priority_train, y_priority_test = train_test_split(
    X_priority,
    y_priority,
    test_size=0.20,
    random_state=42,
    stratify=y_priority
)

# Second split: Training + Validation
X_priority_train_final, X_priority_val, y_priority_train_final, y_priority_val = train_test_split(
    X_priority_train,
    y_priority_train,
    test_size=0.20,
    random_state=42,
    stratify=y_priority_train
)

print("Priority training samples:", len(X_priority_train_final))
print("Priority validation samples:", len(X_priority_val))
print("Priority test samples:", len(X_priority_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

priority_tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=20000,
    ngram_range=(1, 2),
    sublinear_tf=True
)

# Fit ONLY on priority training data
X_priority_train_tfidf = priority_tfidf.fit_transform(
    X_priority_train_final
)

# Transform validation and test data
X_priority_val_tfidf = priority_tfidf.transform(
    X_priority_val
)

X_priority_test_tfidf = priority_tfidf.transform(
    X_priority_test
)

print("Priority training TF-IDF shape:", X_priority_train_tfidf.shape)
print("Priority validation TF-IDF shape:", X_priority_val_tfidf.shape)
print("Priority test TF-IDF shape:", X_priority_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score

priority_lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

priority_lr.fit(
    X_priority_train_tfidf,
    y_priority_train_final
)

priority_lr_pred = priority_lr.predict(
    X_priority_val_tfidf
)

lr_val_accuracy = accuracy_score(
    y_priority_val,
    priority_lr_pred
)

lr_val_macro_f1 = f1_score(
    y_priority_val,
    priority_lr_pred,
    average="macro"
)

print("Priority Logistic Regression")
print("Validation Accuracy:", lr_val_accuracy)
print("Validation Macro F1:", lr_val_macro_f1)
```

```python
from sklearn.svm import LinearSVC

priority_svm = LinearSVC(
    C=1.0,
    random_state=42
)

priority_svm.fit(
    X_priority_train_tfidf,
    y_priority_train_final
)

priority_svm_pred = priority_svm.predict(
    X_priority_val_tfidf
)

svm_val_accuracy = accuracy_score(
    y_priority_val,
    priority_svm_pred
)

svm_val_macro_f1 = f1_score(
    y_priority_val,
    priority_svm_pred,
    average="macro"
)

print("Priority Linear SVM")
print("Validation Accuracy:", svm_val_accuracy)
print("Validation Macro F1:", svm_val_macro_f1)
```

```python
%%writefile app.py

import streamlit as st
import joblib

# Load trained models and TF-IDF vectorizers
queue_model = joblib.load("queue_model.pkl")
queue_tfidf = joblib.load("queue_tfidf.pkl")

priority_model = joblib.load("priority_model.pkl")
priority_tfidf = joblib.load("priority_tfidf.pkl")


# Page configuration
st.set_page_config(
    page_title="AI Customer Support Classifier",
    page_icon="🤖",
    layout="centered"
)

# Title
st.title("🤖 AI Customer Support Ticket Classifier")
st.write(
    "Enter a customer support ticket to predict its "
    "support queue and priority."
)

# Ticket input
ticket = st.text_area(
    "Customer Support Ticket",
    placeholder="Example: My payment was deducted twice..."
)

# Prediction button
if st.button("Predict"):

    if ticket.strip() == "":
        st.warning("Please enter a customer support ticket.")

    else:
        # Queue prediction
        queue_text = queue_tfidf.transform([ticket])
        predicted_queue = queue_model.predict(queue_text)[0]

        # Priority prediction
        priority_text = priority_tfidf.transform([ticket])
        predicted_priority = priority_model.predict(priority_text)[0]

        # Display results
        st.subheader("Prediction Results")

        st.write("📂 **Support Queue:**", predicted_queue)
        st.write("🚨 **Priority:**", predicted_priority.capitalize())
```

```python
%%writefile app.py

import streamlit as st
import joblib

# Load trained models and TF-IDF vectorizers
queue_model = joblib.load("queue_model.pkl")
queue_tfidf = joblib.load("queue_tfidf.pkl")

priority_model = joblib.load("priority_model.pkl")
priority_tfidf = joblib.load("priority_tfidf.pkl")


# Page configuration
st.set_page_config(
    page_title="AI Customer Support Classifier",
    page_icon="🤖",
    layout="centered"
)


# Header
st.title("🤖 AI Customer Support Ticket Classifier")

st.write(
    "Automatically classify customer support tickets "
    "into the appropriate support queue and priority level."
)

st.divider()


# Ticket input
st.subheader("📝 Enter Customer Support Ticket")

ticket = st.text_area(
    "Customer Ticket",
    placeholder=(
        "Example: My credit card was charged twice "
        "for the same purchase."
    ),
    height=180
)


# Prediction
if st.button("🔍 Predict Ticket", use_container_width=True):

    if not ticket.strip():

        st.warning("⚠️ Please enter a customer support ticket.")

    else:

        # Queue prediction
        queue_text = queue_tfidf.transform([ticket])
        predicted_queue = queue_model.predict(queue_text)[0]

        # Priority prediction
        priority_text = priority_tfidf.transform([ticket])
        predicted_priority = priority_model.predict(priority_text)[0]

        # Results
        st.divider()
        st.subheader("📊 Prediction Results")

        col1, col2 = st.columns(2)

        with col1:
            st.success(
                f"📂 **Support Queue**\n\n"
                f"{predicted_queue}"
            )

        with col2:
            st.info(
                f"🚨 **Priority**\n\n"
                f"{predicted_priority.capitalize()}"
            )


# Footer
st.divider()

st.caption(
    "AI-Based Customer Support Ticket Classification & "
    "Priority Prediction | Machine Learning + NLP"
)
```

```python
%%writefile app.py

import streamlit as st
import joblib


# --------------------------------------------------
# Load Models
# --------------------------------------------------

queue_model = joblib.load("queue_model.pkl")
queue_tfidf = joblib.load("queue_tfidf.pkl")

priority_model = joblib.load("priority_model.pkl")
priority_tfidf = joblib.load("priority_tfidf.pkl")


# --------------------------------------------------
# Page Configuration
# --------------------------------------------------

st.set_page_config(
    page_title="AI Customer Support Classifier",
    page_icon="🤖",
    layout="centered"
)


# --------------------------------------------------
# Header
# --------------------------------------------------

st.title("🤖 AI Customer Support Ticket Classifier")

st.write(
    "An NLP-based machine learning system that automatically "
    "classifies customer support tickets into the appropriate "
    "support queue and priority level."
)

st.divider()


# --------------------------------------------------
# Ticket Input
# --------------------------------------------------

st.subheader("📝 Enter Customer Support Ticket")

ticket = st.text_area(
    "Customer Ticket",
    placeholder=(
        "Example: My payment was deducted twice "
        "for the same transaction."
    ),
    height=180
)


# --------------------------------------------------
# Prediction
# --------------------------------------------------

if st.button("🔍 Predict Ticket", use_container_width=True):

    if not ticket.strip():

        st.warning("⚠️ Please enter a customer support ticket.")

    elif len(ticket.strip()) < 10:

        st.warning(
            "⚠️ Please enter a more detailed ticket "
            "for a meaningful prediction."
        )

    else:

        # Queue prediction
        queue_text = queue_tfidf.transform([ticket])
        predicted_queue = queue_model.predict(queue_text)[0]

        # Priority prediction
        priority_text = priority_tfidf.transform([ticket])
        predicted_priority = priority_model.predict(priority_text)[0]

        # --------------------------------------------------
        # Results
        # --------------------------------------------------

        st.divider()

        st.subheader("📊 Prediction Results")

        col1, col2 = st.columns(2)

        with col1:
            st.success(
                f"📂 **Support Queue**\n\n"
                f"{predicted_queue}"
            )

        with col2:

            if predicted_priority == "high":
                st.error(
                    f"🚨 **Priority**\n\n"
                    f"{predicted_priority.capitalize()}"
                )

            elif predicted_priority == "medium":
                st.warning(
                    f"⚠️ **Priority**\n\n"
                    f"{predicted_priority.capitalize()}"
                )

            else:
                st.info(
                    f"🔵 **Priority**\n\n"
                    f"{predicted_priority.capitalize()}"
                )


# --------------------------------------------------
# About Project
# --------------------------------------------------

st.divider()

with st.expander("ℹ️ About This Project"):

    st.write(
        """
        This project uses Natural Language Processing (NLP) and
        Machine Learning to automatically classify customer support
        tickets.

        The system performs two predictions:

        • Support Queue — identifies the appropriate department.

        • Priority — predicts whether the ticket has low, medium,
          or high priority.

        Text data is converted into numerical features using
        TF-IDF, followed by Linear Support Vector Machine (SVM)
        classification.
        """
    )


# --------------------------------------------------
# How It Works
# --------------------------------------------------

with st.expander("⚙️ How It Works"):

    st.write(
        """
        1. Customer enters a support ticket.

        2. The ticket text is converted into TF-IDF features.

        3. The Queue Classification model predicts the support
           department.

        4. The Priority Prediction model predicts the ticket
           priority.

        5. Both predictions are displayed in the application.
        """
    )


# --------------------------------------------------
# Model Information
# --------------------------------------------------

with st.expander("📊 Model Information"):

    st.write("**Queue Classification**")
    st.write("Model: Linear SVM")
    st.write("Test Accuracy: 50.44%")
    st.write("Test Macro F1: 48.50%")

    st.write("")

    st.write("**Priority Prediction**")
    st.write("Model: Linear SVM")
    st.write("Test Accuracy: 59.50%")
    st.write("Test Macro F1: 57.23%")


# --------------------------------------------------
# Footer
# --------------------------------------------------

st.divider()

st.caption(
    "AI-Based Customer Support Ticket Classification & "
    "Priority Prediction | NLP + Machine Learning + Streamlit"
)
```
